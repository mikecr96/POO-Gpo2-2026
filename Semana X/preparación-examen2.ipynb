2. Implementa el método `estudiar_caso(self)`. 
   * Si la energía del cirujano es mayor a 30, imprime: *"Visualizando anatomía en 3D... El Dr. [nombre] ha encontrado una solución atípica"*, y reduce su energía en 15 puntos por el esfuerzo mental.

3. Prueba tus métodos: haz que Shaun estudie un caso y que Melendez tome un café.

```python
class Cirujano:
    medicos_activos = 0

    def __init__(self, nombre:str, especialidad:str, energia:int) -> None:
        self.nombre = nombre
        self.especialidad = especialidad
        self.energia = energia
        self.herramientas = []
        Cirujano.medicos_activos += 1

    def obtener_instrumento(self, instrumento:str):
        self.herramientas.append(instrumento)

    def tomar_cafe(self):
        self.energia += 20
        if self.energia > 100: self.energia = 100
        print(f'La nueva energía del Dr. {self.nombre} es de {self.energia}')

    def estudiar_caso(self):
        if self.energia > 30:
            print(f'Visualizando anatomía en 3D... El Dr. {self.nombre} ha encontrado una solución atípica')
            self.energia -= 15
        else:
            print(f"El Dr. {self.nombre} está bloqueado y necesita ir a la sala de descanso")
            

shaun = Cirujano('Shaun Morphy', 'Cirugía pediátrica', 80)
neil = Cirujano('Neil Meléndez', 'Cardiotorácica', 90)
shaun.obtener_instrumento('Bisturí de juguete')
neil.obtener_instrumento('Estetoscopio')
shaun.estudiar_caso()
neil.tomar_cafe()
```

### 🔪 Fase 5: Quirófano (Tiempo estimado: 10 mins)
Es hora de una cirugía de alto riesgo. Copia la clase de la fase 4 e implementa el método `operar_juntos(self, colega)`.
Este método permite que dos médicos colaboren y debe recibir como argumento a **otro objeto de la misma clase**.
1. La cirugía es extenuante: verifica que **ambos** médicos tengan estrictamente más de 40 puntos de energía. Si alguno no los tiene, imprime *"Cirugía pospuesta, el equipo quirúrgico está agotado"*.
2. Si ambos están en condiciones, la operación procede. El cirujano titular (`self`) pierde 35 puntos de energía por liderar. El cirujano asistente (`colega`) pierde 20 puntos.
3. Imprime el resultado de la operación: *"¡Éxito! El Dr. [nombre_titular] ha salvado al paciente con la asistencia del Dr. [nombre_colega]"*.
4. Prueba la interacción: haz que el Dr. Melendez (titular) opere junto con Shaun (asistente). Imprime la energía restante de ambos para confirmar el desgaste.

```python
class Cirujano:
    medicos_activos = 0

    def __init__(self, nombre:str, especialidad:str, energia:int) -> None:
        self.nombre = nombre
        self.especialidad = especialidad
        self.energia = energia
        self.herramientas = []
        Cirujano.medicos_activos += 1

    def obtener_instrumento(self, instrumento:str):
        self.herramientas.append(instrumento)

    def tomar_cafe(self):
        self.energia += 20
        if self.energia > 100: self.energia = 100
        print(f'La nueva energía del Dr. {self.nombre} es de {self.energia}')

    def estudiar_caso(self):
        if self.energia > 30:
            print(f'Visualizando anatomía en 3D... El Dr. {self.nombre} ha encontrado una solución atípica')
            self.energia -= 15
        else:
            print(f"El Dr. {self.nombre} está bloqueado y necesita ir a la sala de descanso")

    def operar_juntos(self, colega:Cirujano):
        if self.energia > 40 and colega.energia > 40:
            self.energia -= 35
            colega.energia -= 20
            print(f"¡Éxito! El Dr. {self.nombre} ha salvado al paciente con la asistencia del Dr. {colega.nombre}")
        else:
            print("Cirugía pospuesta, el equipo quirúrgico está agotado")

            

shaun = Cirujano('Shaun Morphy', 'Cirugía pediátrica', 80)
neil = Cirujano('Neil Meléndez', 'Cardiotorácica', 90)
shaun.obtener_instrumento('Bisturí de juguete')
neil.obtener_instrumento('Estetoscopio')

neil.operar_juntos(shaun)
print(f'Energía de Neil: {neil.energia}. Energía de Shaun: {shaun.energia}')
```
