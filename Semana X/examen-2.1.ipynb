# 🧟‍♂️ Examen Parcial 2: Protocolo de Supervivencia

**Instrucciones:** La civilización ha colapsado y el campus de la universidad es ahora un refugio. Modelaremos un sistema para gestionar a los sobrevivientes utilizando Programación Orientada a Objetos. 
* **Dificultad:** 4/5 ☠️
* **Duración:** 90 minutos.
* **Nota:** Lee bien cada requerimiento. Deberás implementar **métodos especiales (dunder methods)** de Python para resolver varios de los incisos.

### 🎒 Fase 1: El Refugio y los Sobrevivientes
1. Crea la clase `Sobreviviente`. 
2. Define un atributo de clase llamado `poblacion_refugio` que inicie en `0`.
3. Implementa el método constructor para que cada sobreviviente nazca obligatoriamente con los siguientes atributos de instancia: `nombre` (string), `salud` (entero que inicia en 100), `infeccion` (entero que inicia en 0) y `mochila` (una lista vacía).
4. El constructor debe aumentar en 1 la `poblacion_refugio` cada vez que se instancie un nuevo sobreviviente.

---

### 🪪 Fase 2: Identificación y Conteo
El sistema del refugio necesita poder leer la información de los sobrevivientes fácilmente.
1. Implementa el método especial necesario para que, al usar la función `print()` directamente sobre un objeto de esta clase, devuelva un string exactamente con este formato: 
   *"Sobreviviente: [nombre] | Salud: [salud]% | Infección: [infeccion]%"*
2. Crea un método regular llamado `saquear` que reciba un string (ej. "Lata de frijoles") y lo agregue a la lista de la mochila.
3. Implementa el método especial necesario para que, al usar la función integrada `len()` directamente sobre tu objeto sobreviviente (ej. `len(sobreviviente_1)`), el programa devuelva **la cantidad de objetos** que el sobreviviente tiene actualmente en su mochila.

---

### 🩸 Fase 3: La Horda y la Caída 
Salir del refugio es peligroso.
1. Crea el método `recibir_mordida`. Este método reduce la `salud` del sobreviviente en 40 puntos y aumenta su `infeccion` en 30 puntos.
2. Implementa el método especial (destructor) que se ejecuta automáticamente cuando un objeto es eliminado de la memoria. 
   * Este método debe imprimir: *"ALERTA: [nombre] ha caído y se ha unido a la horda."*
   * Además, debe **restar 1** al atributo de clase `poblacion_refugio`.

---

### ⚖️ Fase 4: Liderazgo
En el refugio, la jerarquía se basa en la resistencia física.
1. Implementa el método especial `__ge__` para poder utilizar el operador `>=` (Mayor o igual que, _greater than or equal to_) entre dos objetos sobrevivientes.
2. La lógica dictamina que el sobreviviente actual (`self`) es "mayor o igual" que el otro sobreviviente **SOLO SI** tiene mayor o igual `salud` **Y** menor o igual `infeccion`. 
3. El método debe retornar `True` si cumple la condición, o `False` de lo contrario.

---

### 💉 Fase 5: El Sacrificio (Interacción entre Objetos)
Un sobreviviente no sobrevive solo.
1. Implementa el método `curar_compañero`. Este método recibe como argumento a **otro objeto de la misma clase**.
2. **Validación 1:** Verifica si el `self` tiene el string `"Botiquín"` guardado en su `mochila`. Si no lo tiene, imprime *"No tienes suministros médicos para ayudar a [nombre_compañero]"*.
3. **Validación 2:** Si sí tiene el botiquín, verifica que la salud del compañero sea estrictamente mayor a 0 y su infección sea menor a 100. Si ya está muerto o infectado al 100%, imprime *"Es demasiado tarde para [nombre_compañero], ya no es humano."*
4. **Acción:** Si tiene el botiquín y el compañero es salvable, elimina el `"Botiquín"` de la mochila del `self`. Luego, aumenta la salud del `compañero` en 50 puntos (tope máximo 100) y reduce su infección a 0. Imprime *"¡[nombre_self] ha salvado a [nombre_compañero]!"*.

### 🧪 Fase 6: Simulación
En tu última celda, escribe el código para probar toda tu clase:
1. Instancia a "Joel" y a "Ellie".
2. Haz que Joel saquee un "Arma" y un "Botiquín".
3. Usa `len()` en Joel para imprimir cuántos objetos tiene en su mochila.
4. Haz que Ellie reciba una mordida.
5. Usa el operador `>=` en un `if` para comparar si Ellie es mayor o igual a Joel, e imprime un mensaje con el resultado.
6. Haz que Joel cure a Ellie.
7. Imprime a Ellie usando `print()` para ver sus stats actualizados.
8. Elimina a Joel del sistema usando `del` para disparar el destructor y comprueba imprimiendo la `poblacion_refugio`.
