```python
class Animal: 
    def __init__(self, edad:int, especie:str) -> None:
        self.edad = edad
        self.especie = especie

    def hablar(self):
        print("El animal habló")
class Perro(Animal): pass
class Gato(Animal): 
    def __init__(self, edad: int, especie: str, tamano:float) -> None:
        # super(Gato, self).__init__(edad, especie)
        super().__init__(edad, especie) # Llamar al __init__ del padre
        self.tamano = tamano
class Vaca(Animal): pass
```
