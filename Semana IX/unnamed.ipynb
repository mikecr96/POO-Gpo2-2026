```python
"""
1. Mochila
2. Artículo

m1 = Mochila()
a1 = Articulo('lapiz')
"""
class Mochila:
    def __init__(self, capacidad) -> None:
        self._articulos = []
        self._capacidad = capacidad
        self.peso_actual = 0

    def agregarArticulos(self, articulo:Articulo):
        if self.peso_actual + articulo._peso < self._capacidad:
            self._articulos.append(articulo)
            self.peso_actual += articulo._peso
            print(f'El artíuclo {articulo._nombre} que pesa {articulo._peso} kg. se metió a la mochila. Nuevo peso disponible: {self.capacidad - self.peso_actual} kg.')
        else:
            print("No cabe")

    def quitarArticulos(self, articulo:Articulo):
        if articulo in self._articulos:
            # if articulo == i:
            self._articulos.remove(articulo)
            self.peso_actual -= articulo._peso
            print(f'El artíuclo {articulo._nombre} fue removido de la mochila. Nuevo peso disponible: {self.capacidad - self.peso_actual} kg.')
        else:
            print(f"El artíuclo {articulo._nombre} no está en la mochila.")
    
    # Setters y getters (viejito) para articulos
    def getArticulos(self):
        return self._articulos
    
    def setArticulos(self, articulo):
        self._articulos.append(articulo)

    # Setters y getters (moderno) para capacidad
    # Getter
    @property # Decoradores
    def capacidad(self):
        return self._capacidad
    
    # Setter
    @capacidad.setter
    def capacidad(self, new_capacidad):
        self._capacidad = new_capacidad

    def __str__(self) -> str:
        if len(self._articulos) == 0:
            return 'Está vacía'
        else:
            contenido = "Esta mochila contiene:\n"
            for articulo in self._articulos:
                contenido += f'Artículo: {articulo._nombre} | Peso: {articulo._peso}\n'
            contenido += f'Esta mochila tiene {len(self._articulos)} artículos.\n'
            contenido += f"Ocupado {self.peso_actual}/{self._capacidad} Kg."
            contenido += f'Peso disponible {self.capacidad - self.peso_actual} Kg.'
            return contenido
    
class Articulo:
    contador = 0
    def __init__(self, nombre:str, peso:float) -> None:
        Articulo.contador += 1
        self._nombre = nombre
        self._peso = peso

    def __str__(self) -> str:
        return f'El artículo {self._nombre} pesa {self._peso} kg.'
```

```python
a1 = Articulo('lápiz', .250)
a2 = Articulo('MacBook', 3)
print(a1, a2)
m1 = Mochila(10)
```
