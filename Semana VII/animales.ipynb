```python
class Animal:
    def __init__(self, nombre, tipo, energia, fuerza:int) -> None:
        self.nombre = nombre
        self.tipo = tipo
        self.energia = energia
        self.fuerza = fuerza

    def isAlive(self):
        return self.energia > 0
    
    def comer(self, comida):
        print(f"{self.nombre} come {comida}")
        self.energia += 10

    def reproducirse(self, pareja:Animal):
        if pareja.isAlive() and self.tipo == pareja.tipo:
            print(f'{self.nombre} y {pareja.nombre} tienen un bebé!')
            return Animal(f'Bebé de {self.nombre}', self.tipo, 20, self.fuerza // 2)
        

animales = [
    Animal('León Leo', 'depredador', 50, 15),
    Animal('Tigre Toño', 'depredador', 45, 14),
    Animal('Zebra Ziggy', 'herbívoro', 30, 5),
    Animal('Elefante Ellie', 'herbívoro', 60, 8),
    Animal('Oso Otto', 'omnívoro', 40, 10),
    Animal('Mono Botas', 'omnívoro', 25, 6)
]

print('Animales en la jungla:')
for animal in animales:
    print(f'{animal.nombre} ({animal.tipo})')

# 3 turnos
```
