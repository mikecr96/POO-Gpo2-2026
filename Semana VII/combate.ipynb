# Juego

## Mapa (100, 100)

### P. Principal (Guerrero)
* nombre
* vida_max
* vida_actual
* ataque
* defensa
### Enemigos (Duendes)
* nombre
* vida_max
* vida_actual
* ataque
* defensa


```python
import random
```

```python
class Personaje:
    def __init__(self, nombre, vida_max, ataque, defensa):
        self.nombre = nombre
        self.vida_max = vida_max
        self.vida_actual = self.vida_max
        self.ataque = ataque
        self.defensa = defensa

    def isAlive(self):
        return self.vida_actual > 0
    
    # No le llamo manualmente
    def recibir_dano(self, dano_entrante):
        dano_real = 0
        if dano_entrante >= 0 and dano_entrante > self.defensa:
            dano_real = dano_entrante - self.defensa
            self.vida_actual -= dano_real
            # vida actual 15 ... daño de 30 .. defensa 10
        # Esto solo es en caso de que la vida actual baje de 0.
        if self.vida_actual < 0:
            self.vida_actual = 0

        print(f'{self.nombre} recibió {dano_real}. Y ahora su vida es de {self.vida_actual}')

    def atacar(self, objetivo:Personaje):
        print(f'{self.nombre} atacó a: {objetivo.nombre}')
        # 50% de que el golpe sea critico
        es_critico = random.choice([True, False])
        if es_critico:
            print("Golpe crítico!")
            # Pegue 30 % más
            dano = self.ataque * 1.3
        else:
            dano = self.ataque
        objetivo.recibir_dano(dano)

        
        
    
    def curar(self, cura):
        self.vida_actual += cura

    def __str__(self):
        return f'''Nombre: {self.nombre}
Vida máxima: {self.vida_max}
Vida actual: {self.vida_actual}
Ataque: {self.ataque}
Defensa: {self.defensa}'''
```

```python
P1 = Personaje('Héroe', 200, 50, 30) 
P2 = Personaje('Duende 1', 90, 60, 50)
P3 = Personaje('Duende 2', 90, 60, 50)
print(P1, P2, P3)
```
