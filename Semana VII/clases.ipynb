```python
class Perro:
    # Métodos especiales
    # Otros lenguajes método constructor. En Python, este es el método iniciador
    def __init__(self, color, nombre, raza, tamano, no_patas=4): 
        # Aquí definimos los atributos de la instacia
        self.color = color
        self.nombre = nombre
        self.raza = raza
        self.tamano = tamano
        self.no_patas = no_patas

    def __str__(self):
        return f"El perro {self.nombre} es de color {self.color}, de raza {self.raza} y de tamaño {self.tamano} y tiene {self.no_patas} patas"
    
    def __len__(self):
        return self.no_patas

fido = Perro('café', 'Fido', 'beagle', 'mediano')
print(fido)
```

```python
coby = Perro('café', 'Coby', 'Dobberman', 'grande')
print(coby)
```

```python
str(coby) # Los dos disparadores del método especial str, son print() y str()
```
