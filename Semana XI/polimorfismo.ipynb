```python
class Vector:
    def __init__(self, x:int, y:int) -> None:
        self.x = x
        self.y = y

    # Sobrecarga de operadores (+)
    def __add__(self, other):
        # Buena praxis es no modificar alguno de los objetos, sino devolver uno nuevo
        return Vector(self.x + other.x, self.y + other.y)
    
    # Sobrecarga de operadores (-)
    def __sub__(self, other):
        # Buena praxis es no modificar alguno de los objetos, sino devolver uno nuevo
        return Vector(self.x - other.x, self.y - other.y)
    
    def __str__(self):
        return f"({self.x}, {self.y})"

v1 = Vector(3, 5)
v2 = Vector(2, 9)
v3 = v1 + v2
print(v3)
v4 = v2 - v1
print(v4)
```

```python
class Fraccion:
    def __init__(self, numerador=0, denominador=1) -> None:
        self.numerador = numerador
        self.denominador = denominador
        if denominador == 0:
            raise ValueError("El denominador no puede ser cero.")
        
    def __str__(self) -> str:
        # Escenario A
        if self.denominador == 1:
            return str(self.numerador)
        # Escenario B
        elif self.denominador != 1:
            return f"{self.numerador} / {self.denominador}"
        # Escenario C
        # En caso de que exista un entero

    def __add__(self, other:Fraccion):
        return Fraccion(self.numerador + other.numerador, self.denominador + other.denominador)
        
f1 = Fraccion(10, 2)
f2 = Fraccion(3, 4)
f3 = f1 + f2
print(f3)
```
