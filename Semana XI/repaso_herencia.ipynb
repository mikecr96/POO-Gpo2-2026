```python
# Clase madre
class Automovil:
  def __init__(self, puertas:int, volante:bool, espejos:int, color:str, year:int, asientos:int) -> None:
    self.puertas = puertas
    self.volante = volante
    self.espejos = espejos
    self.color = color
    self.year = year
    self.asientos = asientos

  def hacer_run_run(self):
    print( "Run run")

  def __str__(self) -> str:
    return f"Este auto tiene {self.puertas} puertas, {self.espejos} espejos, es de color {self.color}, es año {self.year} y tiene {self.asientos} asientos."
```

```python
a1 = Automovil(3, True, 2, "amarillo", 2020, 4)
```

```python
a1.hacer_run_run()
```

```python
print(a1)
```

```python
# Override (reemplazar, sobreescribir) a un método

class SUV(Automovil):
  def hacer_run_run(self):
    print("Ran ran")

suv1 = SUV(5, True, 2, "negra", 2026, 7)
suv1.hacer_run_run()
```

```python
# Añadir funcionalidades a un método

class SUV(Automovil):
  def __init__(self, puertas: int, volante: bool, espejos: int, color: str, year: int, asientos: int, cilindros:int) -> None:
    super().__init__(puertas, volante, espejos, color, year, asientos)
    self.cilindros = cilindros

  def hacer_run_run(self):
    super().hacer_run_run()
    print("Run Run")

suv1 = SUV(5, True, 2, "negra", 2026, 7, 4)
suv1.hacer_run_run()
# suv1.hacer_run_run()
```
