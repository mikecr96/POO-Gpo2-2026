```python
# Superclase /padre / madre
class Animal:
    def ladrar(self):
        print("Guau guau")

    def maullar(self):
        print("miau miau")

    def mugir(self):
        print("muuuuu")

class Perro(Animal): pass
class Gato(Animal): pass
class Vaca(Animal):pass

p1 = Perro()
g1 = Gato()
v1 = Vaca()
animales = [p1, g1, v1]
for animal in animales:
    if type(animal) == Perro:
        animal.ladrar()
    elif type(animal) == Gato:
        animal.maullar()
    elif type(animal) == Vaca:
        animal.mugir()
```

```python
# Superclase /padre / madre
class Animal:
    def hacer_ruido(self):
        print("ruido...")

class Perro(Animal):
    def hacer_ruido(self):
        print("Guau guau")

class Gato(Animal): 
    def hacer_ruido(self):
        print("Miau miau")

class Vaca(Animal):
    def hacer_ruido(self):
        print("Muuuuuu")

p1 = Perro()
g1 = Gato()
v1 = Vaca()
animales = [p1, g1, v1]
for animal in animales:
    animal.hacer_ruido()
```

```python
class Sobreviviente:
    def __init__(self, nombre:str, salud_max:int=100) -> None:
        self.nombre = nombre
        self.salud = salud_max
        self.mochila = []

    def buscar_suministros(self):
        print(f"{self.nombre} está buscando suministros en las ruinas...")

    def defenderse(self):
        print(f"{self.nombre} se defiende con un ataque cuerpo a cuerpo básico.")

    def mostrar_stats(self):
        print(f"[{self.nombre}]\nSalud: {self.salud} | Inventario: {len(self.mochila)}")

miguel = Sobreviviente("Miguel")
miguel.buscar_suministros()
miguel.defenderse()
miguel.mostrar_stats()
```

```python
# La clase médico hereda de Sobreviviente
class Medico(Sobreviviente):
    def __init__(self, nombre: str, salud_max: int = 100, botiquines=3) -> None:
        super().__init__(nombre, salud_max)
        self.botiquines = botiquines

    # Método exclusivo de Médico
    def curar(self, aliado:Sobreviviente):
        if self.botiquines > 0: 
            self.botiquines -= 1
            print(f"{self.nombre} usa un botiquin para curar a {aliado.nombre}.\nBotiquines disponibles: {self.botiquines}")
            aliado.salud = min(100, aliado.salud + 25)
        else: print(f"{self.nombre} no tiene botiquines para curar a {aliado.nombre}")

andy = Medico("Andy", botiquines=1)
andy.buscar_suministros()
andy.defenderse()
andy.mostrar_stats()
andy.curar(miguel)
```

```python
class Soldado(Sobreviviente):
    def __init__(self, nombre: str, salud_max: int = 120, municion=30) -> None:
        super().__init__(nombre, salud_max)
        self.municion = municion

    # Override (sobreescribir) el comportamiento del padre
    def defenderse(self):
        if self.municion > 0:
            self.municion -= 1
            print(f"{self.nombre} usó su arma para defenderse. Munición disponible: {self.municion}")
        else:
            super().defenderse()

ryan = Soldado("Ryan", municion=1)
ryan.buscar_suministros()
ryan.defenderse()
ryan.mostrar_stats()
ryan.defenderse()
```
