```python
from typing import Any


class MochilaJuego:
    def __init__(self, nombre_sobreviviente:str) -> None:
        self._items = []
        self._nombre = nombre_sobreviviente
    # Sobrecar len()
    def __len__(self):
        return len(self._items)
    
    # Sobrecargar in
    def __contains__(self, item):
        return item in self._items
    
    # Sobrecargar for
    def __iter__(self):
        return iter(self._items)
    
    # Sobrecarga objeto()
    def __call__(self) -> Any:
        return f"Esta mochila es de {self._nombre} y tiene {len(self)} elementos"
        
    # Sobrecargar [indexing]
    # def __getitem__(self, key):
    #     return self._items[key]
    
    def meter_objeto(self, objeto:str):
        self._items.append(objeto)
```

```python
mochila = MochilaJuego("Miguel")
# mochila._items # No se hace
objetos_a_meter = ["Cuchillo", "Botiquín", "Machete", "Pañuelos"]
for objeto in objetos_a_meter:
    mochila.meter_objeto(objeto)
print(len(mochila))
"Cuchillo" in mochila
# mochila[0]

for i in mochila:
    print(i)

mochila()
```
