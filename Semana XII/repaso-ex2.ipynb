```python
class Perro:
    def __init__(self, nombre) -> None:
        self._nombre = nombre # Protegido
        self._ladrar = True

    # Getters/Setters para ladrar con @property
    @property
    def ladrar(self): 
        return self._ladrar
    
    @ladrar.setter
    def ladrar(self, ladra:bool):
        self._ladrar = ladra

perro1 = Perro("Firulais")
# perro1.ladrar # Getter
perro1.ladrar = False # setter
perro1.ladrar
```

```python
# Superclase
class Mamifero:
    def __init__(self, nombre:str, edad:int):
        self._nombre = nombre
        self._edad = edad
    
    def hacer_ruido(self):
        print(f"El animal {self._nombre} hizo un ruido.")

class Perro(Mamifero):
    def __init__(self, nombre) -> None:
        self._nombre = nombre # Protegido
        self._ladrar = True

    # Getters/Setters para ladrar con @property
    @property
    def ladrar(self): 
        return self._ladrar
    
    @ladrar.setter
    def ladrar(self, ladra:bool):
        self._ladrar = ladra

# Subclase
class Gato(Mamifero): # Herencia
    # Extender un método
    def __init__(self, nombre: str, edad: int, color:str):
        super().__init__(nombre, edad)
        self.color = color

    # override - sobreescribir
    def hacer_ruido(self):
        print("Miau miau")

    # Crear un método exclusivo de Gato
    def __str__(self) -> str:
        return f"Este gato se llama {self._nombre}, es de color {self.color} y tiene {self._edad} años."

# gato = Gato("Michi", 3, "naranjoso")
# gato.hacer_ruido()
# print(gato)
```

```python
# Polimorfismo
animales = [
    Mamifero("Melman", 1),
    Gato("Silvestre", 3, "negro"),
    Perro("Rambo")
]

for animal in animales: animal.hacer_ruido()
```

```python
class Lapiz:
    def __init__(self, color:str, tamano:str) -> None:
        self.color = color
        self.tamano = tamano

class Mochila:
    def __init__(self) -> None:
        self.items = []

    def agregar_elemento(self, elemento:Lapiz):
        self.items.append(elemento)

lapiz = Lapiz("amarillo", "chico")
mochila = Mochila()
mochila.agregar_elemento(lapiz)
```
