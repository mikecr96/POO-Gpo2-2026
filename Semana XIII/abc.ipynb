# ABSTRACCIÓN

```python
from abc import ABC, abstractmethod
```

```python
# Clase abstracta (La plantilla / El contrato)
class MetodoDePago(ABC):
    def __init__(self, titular) -> None:
        self.titular = titular

    # Este es el método abstracto. No tiene código adentro.
    # Obliga a que cualquier clase que herede de esta, LO TENGA QUE SOBREESCRIBIR (override)
    @abstractmethod
    def procesar_pago(self, monto):
        pass

    def emitir_recibo(self, monto):
        print(f"Recibo generado para {self.titular} por ${monto}")

class PagoConTarjeta(MetodoDePago): 
    # DEBO sobreescribir el método abstracto
    def procesar_pago(self, monto):
        print(f"Se descontaron ${monto} de tu tarjeta")

class PagoConEfectivo(MetodoDePago): 
    # DEBO sobreescribir el método abstracto
    def procesar_pago(self, monto):
        print(f"Recibimos ${monto} en efectivo. Gracias por tu compra")

class PagoMercadoPago(MetodoDePago): 
    # DEBO sobreescribir el método abstracto
    def procesar_pago(self, monto):
        print(f"Procesamos tu pago por ${monto} y lo descontamos de tu wallet de MP")
```

```python
# Intentamos instaciar una subclase SIN implementar el método abstracto
metodo2 = PagoConTarjeta("Diego")
```

```python
metodo2.procesar_pago(100)
```

```python
metodo3 = PagoMercadoPago("Andrea")
```

```python
metodo4 = PagoConEfectivo("Miguel")
```

```python
# Polimorfismo OBLIGADO

for metodo in [metodo2, metodo3, metodo4]:
    metodo.procesar_pago(100)
```

```python
# Importar lo necesario ABC, abstractmethod

# Creamos la clase abstracta (heredando de ABC)

class Vehiculo(ABC):
    # Metodo normal. Todos los vehículos heredan tal cual
    def arrancar_motor(self):
        print("El motor está en marcha. ¡Listos para salir!")

    # Método abstracto. Este método NO tiene implementación aquí
    @abstractmethod
    def mover(self): pass
```

```python
class Coche(Vehiculo):
    # OBLIGATORIO: Implementar el método abstracto
    def mover(self):
        print("El coche está rodando por la carretera sobre sus 4 ruedas.")

class Avion(Vehiculo):
    # OBLIGATORIO: Implementar el método abstracto
    def mover(self):
        print("El avión está volando por el cielo.")
```

```python
# ----- PRUEBAS -----

# Instanciamos nuestras clases
coche = Coche()
avion = Avion()

print("Acciones del coche")
coche.arrancar_motor() # Heredado de la superclase
coche.mover()

print("Acciones del avión")
avion.arrancar_motor() # Heredado de la superclase
avion.mover()
```
