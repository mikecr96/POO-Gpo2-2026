## Setters/Getters

```python
class Persona:
    def __init__(self, nombre, edad):
        self.__nombre = nombre
        self.__edad = edad
    
    # Get para saber el valor actual
    def getNombre(self):
        return self.__nombre
    
    # Set para cambiar el valor actual
    def setNombre(self, nombre):
        self.__nombre = nombre

    def getEdad(self):
        return self.__edad
    
    def setEdad(self, edad):
        self.__edad = edad

    def __digerir(self):
        print(f'{self.__nombre} empezó la digestión')

    def __metabolizar(self):
        print(f'{self.__nombre} comenzó la metabolización')

    def comer(self, comida):
        print(f'{self.__nombre} comió {comida}')
        self.__digerir()
        self.__metabolizar()

# p1 = Persona('Miguel', 100)
# print(p1.getNombre())
# print(p1.getEdad())
# p1.setEdad(20)
# print(p1.getEdad())
# p1.setNombre('Sebastián')
# print(p1.getNombre())
```
