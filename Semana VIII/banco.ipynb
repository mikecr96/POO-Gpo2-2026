```python
class ATM:
    proveedor = "UP"
    fabricante = "China"

    def __init__(self, ubicacion, banco, usuario, saldo, estado, nip):
        self.ubicacion = ubicacion
        self.banco = banco
        self.usuario = usuario
        self.saldo = saldo
        self.estado = estado
        self.nip = nip

    def consultar_saldo(self):
        if self.saldo > 0:
            return self.saldo  
        else:
            self.saldo = 0
            return self.saldo 
        
    def depositar(self, cantidad):
        if cantidad > 0: 
            self.saldo += cantidad

    def transferencia(self, trans, objetivo:ATM):
        if self.saldo > 0 and trans > 0 and self.saldo > trans:
            print(f'{self.usuario} transfirió {trans} a {objetivo.usuario}')
            self.saldo -= trans
            objetivo.saldo += trans
        else:
            print("No se pudo transferir")

            
    """
    __init()__
    consultar saldo
    depósito
    transferencia
    retiro
    cambiar nip
    pago de servicio
    está activa (robada) (vigencia)
    """
```
